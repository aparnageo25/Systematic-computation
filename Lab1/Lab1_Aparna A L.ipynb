```python
#2

vowels = {"a","e","i","o","u"}

def vowel_count(string, vowels):
  vowel_dict= dict.fromkeys(vowels,0)
  for i in string:
    if i in vowels:
      vowel_dict[i]+=1

  if len(vowel_dict)==0:
    print("There is no vowels in the string")

  else:
    return vowel_dict

string = input("Enter a String : ")
vowel_count(string, vowels)
```

```python
#3 prime
def prime(a):
  for i in range (2,a):
    if a%i==0:
      print(f"{a} is not prime number")
      break

  else:
    print(f"{a} is a prime number")


a = int(input("Enter a value greater than 1 :"))
prime(a)
```

```python
#6
def symmetry_palindrom(string):
    a= len(string)//2
    if string[:a]==string[a:]:
      print(f"{string} is a symmetrical word")

    else:
      print(f"{string} is not a symmetrical word")
    if string[0:]== string[::-1]:
      print(f"{string} is a palindrom word")

    else:
     print(f"{string} is not a palindrom word")

string = input("Enter a string : ")
symmetry_palindrom(string)
```

```python
#9
class operations:
  def __init__(self,num1):
    self.num1= num1

  def add(self,num2):
    print(f"sum of the values are {(self.num1+num2)}")

  def subtract(self,num2):
    print(f"Difference of the values are {(self.num1-num2)}")

  def multiply(self,num2):
    print(f"Product of the values are {(self.num1*num2)}")

  def division(self,num2):
    print(f"Product of the values are {(self.num1/num2)}")

b=int(input("Enter first value: "))
c=int(input("Enter second value: "))
a= operations(b)
a.add(c)
a.subtract(c)
a.multiply(c)
a.division(c)
```

```python
#10
def list_sort(my_list):
  even_list = []
  odd_list = []
  for i in my_list:
    if i%2 == 0:
      even_list.append(i)
    else:
      odd_list.append(i)

  print(f"Even Numbers {even_list}")
  print(f"odd Numbers {odd_list}")


my_list= [int(x)for x in input("Enter values seperated by space: ").split()]
list_sort(my_list)
```

```python
#11

def max_marks(marks):
  max=0
  for i in marks.values():
    if i>= max:
      max = i
  subject= list(marks.keys())[list(marks.values()).index(max)]
  print(f"Subject with highest mark is {subject} ")

marks = {
"math": 80,
"english": 75,
"science": 100,
"computer": 95
}

max_marks(marks)
```

```python
#13
def anagram(string1,string2):
  if len(string1)==len(string2):
    for i in string1:
      if i in string2:
        break
    return("Strings are anagram")
  else:
    print("Strings are not anagram")


string1,string2 = input("Enter 2 strings separated by space: ").split()
anagram(string1,string2)
```

```python
#15
def sequence(nums):
  nums.sort()
  longest_sequence= 1
  current_sequence= 1
  for i in range(1,len(nums)):
    if  nums[i] == nums[i-1]:
      continue
    if  nums[i] == nums[i-1]+1:
      current_sequence+=1

    else:
      longest_sequence= max(current_sequence,longest_sequence)
      current_sequence= 1
  return max(current_sequence,longest_sequence)

nums= [int(x)for x in input("Enter values separated by space: ").split()]
sequence(nums)
```
